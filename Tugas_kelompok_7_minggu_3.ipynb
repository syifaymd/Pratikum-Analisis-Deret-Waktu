#Nomor 1

1. Buatlah simulasi deret MA (1) dengan panjang pengamatan 20 dengan nilai parameter $\theta=0.1, \theta=0.3, \theta=0.7, \theta=1.0, \theta=1.5, \theta=2.0$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

np.random.seed(98)

theta_list = [0.1, 0.3, 0.7, 1.0, 1.5, 2.0]

hasil_ma1 = {}

fig, ax = plt.subplots(2, 3, figsize=(15, 7))
ax = ax.flatten()

for j, theta in enumerate(theta_list):

    et = [np.random.normal(0,1)]
    ma1_m = []

    for i in range(20):
        et.append(np.random.normal(0,1))
        ma1_m.append(et[i+1] - theta*et[i])

    hasil_ma1[theta] = np.array(ma1_m)

    ax[j].plot(ma1_m, '-o')
    ax[j].set_title(f'MA(1), θ = {theta}')
    ax[j].set_xlabel('Waktu')
    ax[j].set_ylabel(r'$Y_t$')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Set seed agar hasil random konsisten
np.random.seed(98)

# Parameter MA(1)
theta = 0.25

# Daftar semua panjang pengamatan (N)
lengths = [60, 100, 150, 300]

# Membuat plot waktu (Time Series Plot) untuk semua N
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for i, n in enumerate(lengths):
  # Simulasi deret MA(1): X_t = w_t + theta * w_{t-1}
  w = np.random.normal(0, 1, n + 1)
  x = np.zeros(n)

  for t in range(n):
    if t == 0:
      x[t] = w[t + 1]
    else:
      x[t] = w[t + 1] + theta * w[t]

  # Plot deret waktu
  axes[i].plot(x, color='tab:blue', linewidth=1.2)
  axes[i].set_title(f'Simulasi MA(1) dengan Panjang Pengamatan N = {n}')
  axes[i].set_xlabel('Waktu (t)')
  axes[i].set_ylabel('$Y_t$')
  axes[i].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Set seed agar hasil random konsisten
np.random.seed(98)

# Parameter MA(1) dan daftar semua panjang pengamatan (N)
theta = 0.25
lengths = [60, 100, 150, 300]

# Membuat figur dengan 4 subplot (2x2)
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for i, n in enumerate(lengths):
  # Simulasi deret MA(1) untuk setiap N
  w = np.random.normal(0, 1, n + 1)
  x = np.zeros(n)

  for t in range(n):
    if t == 0:
      x[t] = w[t + 1]
    else:
      x[t] = w[t + 1] + theta * w[t]

  # Membuat lag plot menggunakan pandas pada subplot yang sesuai
  pd.plotting.lag_plot(pd.DataFrame(x), lag=1, ax=axes[i], c='k', alpha=0.6)

  axes[i].set_title(f'Lag Plot MA(1) dengan N = {n}')
  axes[i].set_xlabel(r'$Y_{t-1}$')
  axes[i].set_ylabel(r'$Y_t$')
  axes[i].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from statsmodels.graphics.tsaplots import plot_acf

# Set seed agar hasil random konsisten
np.random.seed(98)

# Parameter MA(1)
theta = 0.25

# Daftar panjang pengamatan (N)
lengths = [60, 100, 150, 300]

# Membuat plot ACF untuk setiap panjang pengamatan
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

simulasi_ma = {}

for i, n in enumerate(lengths):
  # 1. Simulasi deret MA(1): X_t = w_t + theta * w_{t-1}
  # Kita bangkitkan white noise dengan ukuran n + 1 agar indeksnya cukup
  w = np.random.normal(0, 1, n + 1)
  x = np.zeros(n)

  for t in range(n):
    # w[t+1] sebagai error saat ini (w_t) dan w[t] sebagai error masa lalu (w_{t-1})
    # Atau menggunakan konvensi umum: X_t = w_t - theta * w_{t-1} (tergantung definisi tanda,
    # di sini kita gunakan formula umum penulisan: X_t = w_t + theta * w_{t-1})
    if t == 0:
      x[t] = w[t + 1]  # Belum ada w_{t-1} di awal
    else:
      x[t] = w[t + 1] + theta * w[t]

  simulasi_ma[n] = x

  # 2. Plot ACF (Nomor 4)
  plot_acf(x, ax=axes[i], lags=15, alpha=0.05)
  axes[i].set_title(
      f'ACF MA(1) dengan Panjang Pengamatan N = {n} ($\\theta={theta}$)'
  )
  axes[i].set_xlabel('Lag')
  axes[i].set_ylabel('Autokorelasi')
  axes[i].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima_process import ArmaProcess
from statsmodels.graphics.tsaplots import plot_acf

# Set seed sesuai jumlah 2 digit terakhir NIM anggota
np.random.seed(98)

# Parameter MA(2)
models = {
    'a': (0.3, 0.8),
    'b': (-0.3, 0.8),
    'c': (0.3, -0.8),
    'd': (-0.3, -0.8)
}

# Membuat 4 grafik ACF
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, (label, (theta1, theta2)) in zip(axes.flatten(), models.items()):

    # Model MA(2)
    ar = np.array([1])
    ma = np.array([1, theta1, theta2])

    ma_process = ArmaProcess(ar, ma)

    # Simulasi 50 pengamatan
    data = ma_process.generate_sample(nsample=50)

    # Plot ACF
    plot_acf(data, lags=15, zero=False, ax=ax)

    ax.set_title(
        f'ACF MA(2) - {label}: θ1 = {theta1}, θ2 = {theta2}'
    )
    ax.set_xlabel('Lag')
    ax.set_ylabel('Autocorrelation')

plt.tight_layout()
plt.show()
```

7. Buatlah simulasi deret AR (1) 100 pengamatan dengan nilai parameter $\phi=0.3, \phi=0.9, \phi=-0.3, \phi=-0.9$

```python
np.random.seed(98)

phis_7 = [0.3, 0.9, -0.3, -0.9]
n7 = 100
ar1_dict = {}

for phi in phis_7:
    ar = np.array([1, -phi])  # koefisien AR dalam sintaks ArmaProcess = -phi (nilai parameter asli)
    ma = np.array([1])
    ar_obj = ArmaProcess(ar, ma)
    ar1_dict[phi] = ar_obj.generate_sample(nsample=n7)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, phi in zip(axes.flatten(), phis_7):
    ax.plot(ar1_dict[phi])
    ax.set_title(fr'AR(1), $\phi$={phi}')
    ax.set_xlabel('Waktu'); ax.set_ylabel(r'$Y_t$')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima_process import ArmaProcess

# Set seed
np.random.seed(98)

# Parameter AR(2)
models_ar2 = {
    'a': (1.5, -0.5),
    'b': (0.75, -0.25),
    'c': (1.0, -0.25),
    'd': (0.5, -0.75)
}

# Menyimpan hasil simulasi
simulasi_ar2 = {}

for label, (phi1, phi2) in models_ar2.items():

    # Dalam statsmodels:
    # AR = [1, -phi1, -phi2]
    ar = np.array([1, -phi1, -phi2])
    ma = np.array([1])

    ar_process = ArmaProcess(ar, ma)

    # Simulasi 100 pengamatan
    data = ar_process.generate_sample(nsample=100)

    simulasi_ar2[label] = data

# Menampilkan hasil simulasi dalam 4 grafik
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, (label, data) in zip(axes.flatten(), simulasi_ar2.items()):

    phi1, phi2 = models_ar2[label]

    ax.plot(data)
    ax.set_title(
        f'AR(2) - {label}: φ1 = {phi1}, φ2 = {phi2}'
    )
    ax.set_xlabel('Waktu')
    ax.set_ylabel(r'$Y_t$')

plt.tight_layout()
plt.show()
```
